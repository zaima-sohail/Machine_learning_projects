```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv(
    "ecommerce_sales_analytics_5000.csv",
    parse_dates=['order_date']
)
df=df.sample(n=100)

               
```

```python
montly_revenue=(
    df
    .set_index("order_date")["revenue"]
    .resample("ME")
    .sum()
)
```

```python
fig ,ax=plt.subplots(figsize=(7,8))
ax.plot(montly_revenue.index, montly_revenue.values,linewidth=2)
ax.set_title("montly revenue trend")
ax.set_xlabel("montly_revenue")
ax.set_ylabel("sales")
ax.legend()
plt.show()
```

```python
fig ,ax=plt.subplots(figsize=(6,7))
ax.scatter(df['unit_price'],df['revenue'],alpha=0.5)

ax.set_title("unitprice vs price")
ax.set_xlabel("unit price")
ax.set_ylabel("revenue")
plt.show()
```

```python
fig ,ax=plt.subplots(figsize=(7,5))
ax.scatter(df["unit_price"],df["revenue"],alpha=0.7)
ax.set_title("unit price vs revenue")
ax.set_xlabel("unit_price")
ax.set_ylabel("revenue")
```

```python
category_name=df.groupby("product_category")["revenue"].sum()
fig ,ax=plt.subplots()
ax.bar(category_name.index,category_name.values)
ax.set_title("product vs revenue")
ax.set_xlabel("products")
ax.set_ylabel("revenue")
```

```python
grouped=df.groupby(["product_category","region"])["revenue"].sum().unstack()
grouped.plot(kind="bar",figsize=(10,4))
plt.title("product category vs revenue")
plt.xlabel("product category,region")
plt.ylabel("revenue total")
plt.show()
```

```python
grouped.plot(kind="bar",stacked=True,figsize=(10,18))
plt.title("stacked plot by category and revenue")
plt.xlabel("proucts")
plt.ylabel("revenue")
plt.show()
```
