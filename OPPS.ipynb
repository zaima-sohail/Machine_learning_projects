```python
class student:
    def __init__(self,name,age,Grade):
        self.name=name
        self.age=age
        self.Grade=Grade
    def details(self):
        print(f"Name of Student is:{self.name}")
        print(f"Age of Student is:{self.age}")
        print(f"Grade of Student is:{self.Grade}")
    def is_eligible(self):
        if self.age>=15:
            print(self.name ,"is eligible to admission")
        else:
            print(self.name ,"not eligible to admission")
    def is_grade(self):
        if self.Grade=="A+":
            print("you are elgible to scholarship")
        else:
            print("you are not eligible to scholarship")
my_student=student("Zaima Sohail",21,"A+")
my_student.details()
my_student.is_eligible()
my_student.is_grade()

my_student1=student("Zoya Sohail",15,"A+")
my_student1.details()
my_student1.is_eligible()
my_student1.is_grade()

my_student2=student("Rida Sohail",21,"A")
my_student2.details()
my_student2.is_eligible()
my_student2.is_grade()
```

```python
class Student:
    def __init__(self,name,age):
        self.__age=age
        self.name=name
    def info(self):
       print(f"Name of Student is:{self.name}")
       print(f"age of student is:{self.__age}")
    def get__age(self):
        return self.get__age
    
students=Student("Zaima",21)
students.info()
print(students.get__age())
```

```python
class Bank:
    def __init__(self,amount,balance):
        self.__balance=balance
        self.amount=amount
    def deposit(self):
        self.amount+=self.__balance
        print(f"the balance is:{self.amount}")
        print(f"the amount after is:{self.__balance}")
    def get__balance(self):
        return self.__balance
emp=Bank(1000,5000)
emp.deposit()
print(emp.get__balance())
        
```

```python
class Result:
    def __init__(self,name,age,grade):
        self.name=name
        self.age=age
        self.__grade=grade
    def infos(self):
        print(f"Name of student is:{self.name}")
        print(f"Age of student is:{self.age}")
        print(f"Grade of  student is:{self.__grade}")
    
    def set_grade(self):
         if self.__grade=="A+":
             print("excellent !")
         elif self.__grade=="B+":
             print("Good")
         elif self.__grade=="C":
             print("improve")
         else:
            print("you need to work hard")
    def get_grade(self):
        return self.__grade
stu=Result("Zaima",21,"A+")
stu.infos()
stu.set_grade()
print(stu.get_grade())
             

        
```

```python
class Animal:
    def sound(self):
        print("all animals are different sounds")
class Dog(Animal):
    def Bark(self):
        print("dogs are barking ")
class Cat(Animal):
    def Meow(self):
        print("Cats are cute and sounds meow meow")
dog=Dog()
dog.sound()
dog.Bark()


cat=Cat()
cat.sound()

cat.Meow()
```

```python
class calculator:
    def add(self,a, b=0,c=0):
        return a+b+c
cal1=calculator()
print(cal1.add(5))
print(cal1.add(10,5))
print(cal1.add(10,5,5))
```

```python
class car:
    def speed(self):
        print("the car is moving fast")
class bike:
    def speed(self):
        print("the speed of bike is slower than car")
class cycle:
    def speed(self):
        print("the speed of cycle is much slower than car and bike")
for vehicles in [car(),bike(),cycle()]:
    vehicles.speed()

```
