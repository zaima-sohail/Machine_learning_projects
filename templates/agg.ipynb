```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
from scipy.cluster.hierarchy import dendrogram,fcluster,linkage
from sklearn.cluster import AgglomerativeClustering
```

```python
dataset=load_iris()
X=dataset.data
df=pd.DataFrame(X,columns=dataset.feature_names)
print("dataset of 5 rows:",X.shape)
```

```python
df.head()
```

```python
df.tail()
```

```python

df.isnull().sum()
```

```python
model=AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)
df['agglomerative_analysis']=model.fit_predict(X)
```

```python
df.head()
```

```python
linked=linkage(X,method='ward')
plt.figure(figsize=(10,15))
dendrogram(linked)
plt.title("dendogram of iris dataset")
plt.xticks(rotation=34)
plt.show()
```

```python
plt.figure(figsize=(9,7))
plt.scatter(
    df['petal length (cm)'],
    df['petal width (cm)'],
    s=df['sepal length (cm)'] * 100,
    c=df['agglomerative_analysis'],
    cmap='viridis',
    alpha=0.8
)
plt.title('Agglomerative clustering on Iris data')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.show()
```
