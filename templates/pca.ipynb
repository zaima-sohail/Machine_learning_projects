```python
import numpy as np
```

```python
arr=np.array(
    [
        [5,10],
        [3,7],
        [10,15],
        [2,4]
    ]
)
X_centered=arr-np.mean(arr,axis=0)
(X_centered)
```

```python
Covarinace_matrix=np.cov(X_centered,rowvar=False)
(Covarinace_matrix)
```

```python
eigenvalues,eigenvectors=np.linalg.eig(Covarinace_matrix)
(eigenvalues)
```

```python
idx=np.argsort(eigenvalues)[::1]
eigenvalues=eigenvalues[idx]
eigenvectors=eigenvectors[:,idx]
(eigenvectors)
```

```python
explained_varinace=eigenvalues/np.sum(eigenvalues)
(explained_varinace)
```

```python
import numpy as np 
from sklearn.datasets import load_wine
from sklearn.decomposition import PCA
```

```python
data=load_wine()
```

```python
X=data.data
(X)
```

```python
feature=data.feature_names
(feature)
```

```python
pca=PCA(n_components=3)
X_pca=pca.fit_transform(X)
```

```python
print("the original data of X is:",X.shape)
print("after pca is:",X_pca.shape)
```

```python
explained_varinace=pca.explained_variance_ratio_
(explained_varinace)
```

```python
total_variance=np.sum(explained_varinace)
(total_variance)
```

```python
information=1-(total_variance)
(information)
```

```python
import numpy as np
from  sklearn.datasets import load_breast_cancer
```

```python
data=load_breast_cancer()
X=data.data
feature_name=data.feature_names
(feature_name)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
print("the original data of X is:",X.shape)
print("after scaling the data is:",X_scaled.shape)
```

```python
pca=PCA(n_components=2)
X_pca=pca.fit_transform(X_scaled)
```

```python
print("after scaling the data is:",X_pca.shape)
```

```python
explained_varinace=pca.explained_variance_ratio_
(explained_varinace)
```

```python
print("the total variance is:",np.sum(explained_varinace))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,5))
plt.scatter(
    X_pca[:,0],
    X_pca[:,1],
    c=data.target,
    cmap="plasma")
plt.title("PCA of breast cancer dataset")
plt.xlabel("first principal component")
plt.show()
```
