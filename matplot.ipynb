```python
import matplotlib.pyplot as plt
```

```python
days=[1,2,3,4,5,6,7]
sales=[150,120,130,170,200,220,221]
plt.plot(
    days,sales,
    marker='o',
    linewidth=2,
    alpha=0.8,
    label="sales plot graph")
plt.xlabel("Days")
plt.ylabel("sales")
plt.title("sales trend over week")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
temperature=[20,22,25,27,30,32,35]
sales=[50,65,80,90,120,140,160]
plt.scatter(
    temperature,sales,
    alpha=0.7,
    marker='o',
    label="ice-cream sales vs temperature")
plt.xlabel("Temperature(oC)")
plt.ylabel("Ice cream sales")
plt.legend()
plt.show()

```
