```python
import pandas as pd
```

```python
data={
    'Name':['Alice','bob','charles','David'],
    'Age':[12,34,56,23],
    'Salary':[10000,20000,45000,4700],
    'City':['Lahore','KPK','Karachi','Islamanbad']
}
df=pd.DataFrame(data)
print("ndata framing are:",df)
```

```python
subset_loc=df.loc[1:3,['Name','Salary']]
print("slicing of subset_loc:",subset_loc)
```

```python
subset=df.iloc[0:2,1:3]
print("indexing position:",subset)
```

```python
high_salary=df[df['Salary']>7500]
print("highest salary:",high_salary)
```
